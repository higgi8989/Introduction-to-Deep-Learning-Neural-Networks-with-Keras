```python
import numpy as np

weights = np.around(np.random.uniform(size=6), decimals=2) #creating random weights for each nodes
biases = np.around(np.random.uniform(size=3), decimals=2) #creating biases for each nodes
```

```python
print(weights)
print(biases)
```

```python
#intializing input variables
x_1 = 0.5
x_2 = 0.85

print("x_1 is {} and x_2 is {}".format(x_1, x_2))
```

```python
#Let's compute first node result
z_11 = x_1*weights[0] + x_2*weights[1] + biases[0]

print("z_11 is {}".format(z_11))
```

```python
x1 = 0.3
x2 = 0.7

w1 = 0.2
w2 = 0.4

b = 0.1

z = x1*w1 + x2*w2 + b

z = -4

a1 = 1/(1+ np.exp(-z))

print("z is {} and a1 is {}".format(z, a1))
```

```python
#Let's compute second node of first layer result
z_12 = x_1*weights[2] + x_2*weights[3] + biases[1]

print("z_12 is {}".format(z_12))
```

```python
#calculating sigmoid function for z_11 & z_12

a_11 = 1/(1+np.exp(-z_11))
a_12 = 1/(1+np.exp(-z_12))

print("a_11 is {} and a_12 is {}".format(np.around(a_11, decimals=4), np.around(a_12, decimals=4)))
```

```python
z_2 = a_11*weights[4] + a_12*weights[5] + biases[2]

print("z_2 is {}".format(np.around(z_2, decimals=4)))
```

```python
a_2 = 1/(1 + np.exp(-z_2)) 

print("a_2 is {}".format(np.around(a_2, decimals=4)))

```
